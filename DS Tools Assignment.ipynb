```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#importing libraries
```

```python
file_path = r'paris_listings_analysis.csv'

df = pd.read_csv(file_path) 

df
#read the csv file
```

```python
df.info
```

```python
df.isnull().sum()

#how many nulls
```

```python

df['host_since'] = pd.to_datetime(df['host_since'])


paris_df = df[df['city'] == 'Paris'][['host_since', 'neighbourhood', 'city', 'accommodates', 'price']]


paris_df

#filter only paris
```

```python

paris_df.describe()


```

```python

paris_listings_neighbourhood = (
    paris_df.groupby("neighbourhood").agg({"price": "mean"}).sort_values("price")
)
paris_listings_neighbourhood
#Group by neighbourhood showing its avg prices by mean
```

**Most Expensive Hood** 

```python
most_expensive_hood = paris_listings_neighbourhood.index[-1]
most_expensive_hood

#show the most expensive hood by printing the last record
```

```python
paris_listings_accommodations = (
    paris_df[paris_df['neighbourhood'] == most_expensive_hood].groupby("accommodates").agg({"price": "mean"}).sort_values("price")
)

paris_listings_accommodations

#Group by accommodations just like the neighbourhood phase
```

```python

paris_listings_over_time = (
    paris_df.set_index("host_since").resample("YE").agg({"price": "mean", "city": "count"}).rename(columns={"city": "new_hosts"})
)
paris_listings_over_time


#show new hosts and prices with dates
```

**Visualisation**

```python
sns.set_style("whitegrid")

plt.figure(figsize=(10, 8))
paris_listings_neighbourhood['price'].plot(kind='barh', color='blue')
plt.title("Average Price by Paris Neighbourhood")
plt.xlabel("Price ($)")
plt.ylabel("Neighbourhood")
plt.tight_layout()
plt.show()

#average price chart with every neighbourhood
```

```python

plt.figure(figsize=(8, 6))
paris_listings_accommodations['price'].plot(kind='barh', color='red')
plt.title(f"Price vs Capacity in {most_expensive_hood}")
plt.xlabel("Average Price ($)")
plt.ylabel("Accommodates (Number of People)")
plt.tight_layout()
plt.show()

#price VS Capacity in the most expensive neighbourhood
```

```python

fig, ax1 = plt.subplots(figsize=(12, 6))

ax1.set_xlabel('Year')
ax1.set_ylabel('New Hosts', color='tab:blue')
ax1.plot(paris_listings_over_time.index, paris_listings_over_time['new_hosts'], color='tab:blue', linewidth=2, label='New Hosts')
ax1.tick_params(axis='y', labelcolor='tab:blue')
ax1.set_ylim(0)

ax2 = ax1.twinx()
ax2.set_ylabel('Average Price ($)', color='tab:red')
ax2.plot(paris_listings_over_time.index, paris_listings_over_time['price'], color='tab:red', linestyle='--', linewidth=2, label='Avg Price')
ax2.tick_params(axis='y', labelcolor='tab:red')
ax2.set_ylim(0)

plt.title("Impact of Regulations (2015) on Paris AirBnB Market")
fig.tight_layout()
plt.show()


# final answer what happened after 2015 ? 
```

**WE can conclude that Elysee is the highest average listing price and new hosts have been declined after 2015 while prices have been raised a bit**
